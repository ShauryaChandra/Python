```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression     # used for binary classification (e.g., yes/no, spam/not spam)
from sklearn.model_selection import train_test_split    
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Titanic-Dataset.csv')

display(df.shape)
display(df.head(10))
display(df.isnull().sum())
```

```python
df.drop(columns = ['PassengerId', 'Name', 'Cabin', 'Ticket'], inplace = True)

df['Age'].fillna(df['Age'].median(), inplace = True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace = True)

df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})     # Convert male/female to numbers as ML can't understand texts
df = pd.get_dummies(df, columns = ['Embarked'], drop_first = True)

display(df.isnull().sum())
display(df.shape)
display(df.head(10))
```

```python
x = df.drop(columns = ['Survived'])
y = df['Survived']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

model = LogisticRegression(max_iter = 1000)
model.fit(x_train, y_train)     # to train the model

display("Training done")
display("Train size:", x_train.shape)
display("Test size:", y_test.shape)
```

```python
y_pred = model.predict(x_test)      # asks who survived for these 179 passangers you've never seen before

display("Accuracy:", accuracy_score(y_test, y_pred))
display(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Did not survive', 'Survived'],
            yticklabels=['Did not survive', 'Survived'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Titanic Survival - Confusion Matrix')
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150)
plt.show()
```

```python
importance = pd.Series(model.coef_[0], index=x.columns)
importance = importance.sort_values()

plt.figure(figsize=(8,5))
importance.plot(kind='barh', color='steelblue')
plt.title('Feature Coefficients - Logistic Regression')
plt.xlabel('Coefficient Value')
plt.tight_layout()
plt.savefig('feature_importance.png', dpi=150)
plt.show()
```
