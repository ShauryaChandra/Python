```python
import yfinance as yf
import pandas as pd

df = yf.download("AAPL", start = "2025-01-01", end = "2025-12-31")
display(df.head())
display(df.shape)
```

```python
import matplotlib.pyplot as plt

plt.plot(df.index, df["Close"])
plt.title("AAPL closing price of 2025")
plt.xlabel("Date")
plt.ylabel("price (USD)")
plt.savefig("AAPL_Price.png")
plt.show()
```

```python
# calculate daily returns

print(df.columns)
print(df.head())
df = yf.download("AAPL", start="2023-01-01", end="2024-01-01")
df.columns = df.columns.get_level_values(0)


df["Return"] = (df["Close"] - df["Close"].shift(1)) / df["Close"].shift(1) * 100
display(df["Return"].head(10))
```

```python
#plotting returns as histogram

plt.hist(df["Return"].dropna(), bins = 50, edgecolor = "black")
plt.title("AAPL daily returns")
plt.xlabel("Daily Return %")
plt.ylabel("Frequency")
plt.savefig("aapl_returns.png")
plt.show()
```

```python
tickers = ["AAPL", "TSLA", "MSFT"]

prices = yf.download(tickers, start = "2025-01-01", end = "2025-12-31")

for ticker in tickers:
    plt.plot(prices.index, prices["Close"][ticker], label = ticker)

plt.title("Stock Prices 2025")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend()
plt.savefig("three_stocks.png")
plt.show()
```

```python
## DAY 13 - Rolling averages and Resampling

# Rolling averages: A rolling average smooths noisy stock prices. Helps identify trends, momentum and support/resistance
import yfinance as yf
import matplotlib.pyplot as plt

df = yf.download(
    "AAPL",
    start="2025-01-01",
    end="2025-12-31"
)
df["MA7"] = df["Close"].rolling(window = 7).mean()      # MA = moving average
df["MA30"] = df["Close"].rolling(window=30).mean()
display(df[["Close", "MA7", "MA30"]].head(40))
```

```python
plt.plot(df.index, df["Close"], label = "Actual Price", alpha = 0.5)
plt.plot(df.index, df["MA7"], label = "7 day MA")
plt.plot(df.index, df["MA30"], label = "30 day MA")

plt.title("AAPL prices with Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price(USD)")
plt.legend()
plt.savefig("MovingAverages.png")
plt.show()
```

```python
# Resampling to monthly

monthly = df["Close"].resample("ME").mean()     # ME = month end, W = weekly, YE = year end
display(monthly)

plt.plot(monthly.index, monthly.values, marker = "o")
plt.title("AAPL Monthly Average Price")
plt.xlabel("Month")
plt.ylabel("Price (USD)")
plt.savefig("monthly_price.png")
plt.show()
```
